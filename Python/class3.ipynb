1)	Check for Age eligibility for casting a vote . #only 18+ can vote

```python
name =input("Your Name: ")
age =int(input("Provide your age: "))

if age >= 18 :
    print(f"{name} your are eligible to vote with {age}")
else:
    print(f"{name} you are {age} thereby you are not eligible to vote, wait for {18 - age} more years")
```

Since you want the user to enter the subject names first and then their marks, you can use a dictionary to store the data and then check whether the student has passed or failed.

# Taking subject names and marks from the user

subjects = {}

for i in range(3):
    subject_name = input(f"Enter name of subject {i+1}: ")

    while True:
        marks = int(input(f"Enter marks obtained in {subject_name} (0-100): "))

        if 0 <= marks <= 100:
            subjects[subject_name] = marks
            break
        else:
            print("Marks must be between 0 and 100. Please try again.")

# Checking pass/fail
passed = True

for subject, marks in subjects.items():
    if marks <= 45:  # passing marks should be above 45
        passed = False
        print(f"Failed in {subject} with {marks} marks.")

print("\nSubjects and Marks:")
for subject, marks in subjects.items():
    print(f"{subject}: {marks}")

if passed:
    print("\nResult: PASS")
else:
    print("\nResult: FAIL")
