```python
#Write a program that checks if a number is positive, negative, or zero.

num= int(input("Input a number to check:  "))
if num >= 0:
    print(f"{num} is positive")
else:
    print(f"{num} is negative")
```

```python
#Check whether a given number is even or odd

num= int(input("Input a number to check:  "))

if num % 2==0:
    print(f"{num} is even")
else:
    print(f"{num} is odd")
```

```python
#Take a number as input and print "Eligible to vote" if the age is 18 or above, otherwise print "Not eligible".
name =input("Your Name: ")
age =int(input("Provide your age: "))

if age >= 18 :
    print(f"{name} your are eligible to vote with {age}")
else:
    print(f"{name} you are {age} thereby you are not eligible to vote, wait for {18 - age} more years")
```

```python
#Write a program to find the largest of two numbers using if-else.

num1 = int(input("Please type your number 1: "))
num2 = int(input("Please type your number 2: "))

if num1 > num2:
    print(f"{num1} is greater than {num2} by {num1 - num2}")
else:
    print(f"{num2} is greater than {num1} by {num2 - num1}")
```

```python
#Write a program that checks if a year is a leap year or not.

year =int(input("input year you wish to check: "))

if (year % 4 == 0 and year % 100 != 0):
    print(f"{year} is a leap year")
elif year % 100 == 0 and year % 400 == 0:
    print(" is a century but not leap year")
else:
    print(f"{year} is not a leap year")
```

```python
#Check whether a number is divisible by 5 and 11 or not.
num = int(input("input number to check if divisible by 5 or 11: "))

if (num % 5==0) & (num % 11 == 0) :
    print(f"{num} is divisible by 5 and 11")

elif num % 5== 0:
    print(f"{num} is only divisible by 5")
elif num %11==0:
    print(f"{num} is divisble by only 11")
else:
    print(f"{num} is not divisible by both 5 and 11")
```

```python
#Take three numbers as input and print the largest number among them (using if-elif-else).

num1=int(input("input a number 1: "))
num2=int(input("input a number 2: "))
num3=int(input("input a number 3: "))

if num1>num2 and num1>num3:
    print(f"{num1} is greatest")
elif num2>num1 and num2>num3:
    print(f"{num2} is greatest")
elif num1==num2==num3:
    print("all are same ",num1)
else:
    print(f"{num3} is the greatest")
```

```python
#another take of the same question
#Take three numbers as input and print the largest number among them (using if-elif-else).

num1=int(input("input a number 1: "))
num2=int(input("input a number 2: "))
num3=int(input("input a number 3: "))

if num1>num2 and num1>num3:
    if num2 > num3:
        print(f"{num1} is greatest then {num2} then {num3}")
    else:
        print(f"{num1} is greatest then {num3} then {num2}")
elif num2>num1 and num2>num3:
    if num1 > num3:
        print(f"{num2} is greatest then {num1} then {num3}")
    else:
        print(f"{num2} is greatest then {num3} then {num1}")
elif num1==num2==num3:
    print("all are same ",num1)
else:
    if num2 > num1:
        print(f"{num3} is greatest then {num2} then {num1}")
    else:
        print(f"{num3} is greatest then {num1} then {num2}")
```
