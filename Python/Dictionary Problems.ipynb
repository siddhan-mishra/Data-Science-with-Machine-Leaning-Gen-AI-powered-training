```python
# create a dictonary student with student = {"name":"Rahul","age":20,"course":"Python"}
student = {"name":"Rahul","age":20,"course":"Python"}
```

```python
# print value of age and course from dictionary
print(student["age"])
print(student["course"])
```

```python
# add grade:A to the dictionary a new key-value pair
student["grade"] = "A"
print(student)
```

```python
# update age = 21
student["age"]=21
print(student["age"])
```

```python
#Remove course key from dict
student.pop("course")
print(student)
#can also be done using del student["course"]
# the only thing is in pop we can store the course in another dictionary but not for del
```

```python
#print all the keys and all values of dictionary separately

for x,y in student.items():
    print(x,y)

# also can be done by
# for x ins student:
#     print(x,student[x])
```

```python
# check if name exists in dict 
print(student.get("name","Not there"))
```
