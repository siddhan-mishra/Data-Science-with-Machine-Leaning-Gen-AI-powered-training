5)	check if a statement has passed or failed , by taking marks in 3 subjects . # lets take marks of subjects ( maths , physics , chemistry ) marks are between 0 - 100 and passing marks should be above 45

```python
subject_marks = {}
passing_mark = 45

num = int(input("Please specify no of subjects: "))

for i in range(num):
    subject_name = input(f"Enter the name of subject {i}: ")
    while True:
        try:
            marks = float(input(f"Enter marks obtained in {subject_name}: "))
            if 0 <= marks <= 100:
                break
            else:
                print("Invalid marks!...")
        except ValueError:
            print("Invalid input!...")
subject_marks[subject_name] = marks

student_passed = True

for subject, marks in subject_marks.items():
    if marks > passing_mark:
        print(f"{subject}: {marks} - Passed")
    else:
        print(f"{subject}: {marks} - Failed")
        student_passed = False

if student_passed:
    print("\nOverall Result: PASSED")
else:
    print("\nOverall Result: FAILED")
```

11) Write a Python program to check whether a year is a leap year or not. 

```python
year =int(input("input year you wish to check: "))

if (year % 4 == 0 and year % 100 != 0):
    print(f"{year} is a leap year")
elif year % 100 == 0 and year % 400 == 0:
    print(" is a century but not leap year")
else:
    print(f"{year} is not a leap year")
```
